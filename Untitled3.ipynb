```python
import pandas as pd
#load your excel file
df = pd.read_excel('Share Dataset for Data Analytics2.xlsx')
# sho a quick success message and the first 3 rows
print("Success! Your file is loaded.")
display(df.head(3)) 
```

```python
 # Check howing many missing values are in each column
print("Missing values per column:")
print(df.isnull().sum())
```

```python
# Fill missing couponcode values with "No Coupon"
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')

# Verify that the missing values are now gone
print("Missing values after filling CouponCode:")
print(df['CouponCode'].isnull().sum())
```

```python
# Check how many duplicate rows exist in the dataset
duplicates_count = df.duplicated().sum()
print("Number of duplicate rows found:", duplicates_count)
```

```python
# Check data types and ensure everything is formatted correctly
df.info()
```

```python
# Check unique values in OrderStatus to spot typos or capitalization differences
print(df['OrderStatus'].value_counts()) 
```

```python
# Check unique values in PaymentMethod
print(df['PaymentMethod'].value_counts()) 
```

```python
# Save your clean dataset to a new file
df. to_excel('Cleaned_Share_Dataset.xlsx', index=False)
print("Your cleaned dataset has been saved succcessfully!")
```
